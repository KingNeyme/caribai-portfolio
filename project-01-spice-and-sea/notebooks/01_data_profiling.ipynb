```python
# How many distinct items are there?
pos24["item"].nunique()
```

```python
sorted(pos24["item"].unique())
```

A count was made of the unique items and compared to the items on the `menu.csv` file to see if the numbers matched. The data shows that there are 47 unique items. But it does not match the total items within the `menu.csv` file which has a total of **26**.
